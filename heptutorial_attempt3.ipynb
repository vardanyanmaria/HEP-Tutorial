```python
import ROOT as r

class MyMuon(r.TLorentzVector):
    def __init__(self, px=0, py=0, pz=0, e=0, iso=0.0, charge=0):
        super().__init__(px, py, pz, e)
        self.isolation = iso
        self.charge = charge

    def IsIsolated(self, relcut=0.1):
        if self.Pt() == 0:
            return False
        return (self.isolation / self.Pt()) < relcut


f = r.TFile("files_for_python\data.root")
tree = f.Get("Events")

h_NMuon = r.TH1F("NMuon", "Number of muons", 7, 0, 7)
h_Mmumu = r.TH1F("Mmumu", "Invariant di-muon mass", 60, 60, 120)

MuonRelIsoCut = 0.1
MuonPtCut = 25
EventWeight = 1.0
triggerIsoMu24 = True


for i in range(10): 
    tree.GetEntry(i)    
    Muons = []
    for j in range(tree.NMuon):
        mu = MyMuon(
            tree.Muon_Px[j],
            tree.Muon_Py[j],
            tree.Muon_Pz[j],
            tree.Muon_E[j],
            tree.Muon_Iso[j],
            tree.Muon_Charge[j]
        )
        Muons.append(mu)

    iso_muons = [m for m in Muons if m.IsIsolated(MuonRelIsoCut)]
    N_IsoMuon = len(iso_muons)
    h_NMuon.Fill(N_IsoMuon, EventWeight)

    if N_IsoMuon > 1 and triggerIsoMu24:
        if iso_muons[0].Pt() > MuonPtCut:
            h_Mmumu.Fill((iso_muons[0] + iso_muons[1]).M(), EventWeight)

input("Press Enter to exit...")
```
